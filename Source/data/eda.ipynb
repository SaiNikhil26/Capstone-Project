```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("azraimohamad/coursera-course-data")

print("Path to dataset files:", path)
```

```python
import pandas as pd

df = pd.read_csv("coursera_course_dataset_v3.csv")
```

```python
df.head()
```

```python
# Get the column names
print("Original columns:", list(df.columns))
```

```python
# Check which columns have null values
null_counts = df.isnull().sum()
print("\nNull value counts per column:")
print(null_counts)
```

```python
# Now place zeros in the null values of the course_students_enrolled column
df["course_students_enrolled"] = df["course_students_enrolled"].fillna(0)
```

```python
# Check which columns have null values
null_counts = df.isnull().sum()
print("\nNull value counts per column:")
print(null_counts)
```

```python
# Drop rows where course_description is null
df = df.dropna(subset=["course_description"])
```

```python
# Check which columns have null values
null_counts = df.isnull().sum()
print("\nNull value counts per column:")
print(null_counts)
```

```python
# save this as a new CSV file
df.to_csv("coursera_course_dataset_v3_cleaned.csv", index=False)
```

```python
# count the number of rows
num_rows = len(df)
print(f"Number of rows in the cleaned dataset: {num_rows}")
```
